```python
import pandas as pd
import numpy as np

# Configuración
n_filas = 100000

# Creamos fechas aleatorias entre 2015 y 2026
fechas = pd.to_datetime(np.random.choice(pd.date_range('2015-01-01', '2026-12-31'), n_filas))

# Ventas aleatorias con algunos valores nulos (para que sea "real")
ventas = np.random.uniform(10.0, 500.0, n_filas)
ventas[np.random.choice(n_filas, 500)] = np.nan  # Simulamos 500 errores de registro

# Nuevas columnas categóricas
ciudades = np.random.choice(['Madrid', 'Paris', 'Barcelona', 'Roma', 'Sevilla'], n_filas)
categorias = np.random.choice(['Electrónica', 'Ropa', 'Alimentos', 'Hogar', 'Juguetes'], n_filas)

# Crear el DataFrame
df_gigante = pd.DataFrame({
    'fecha': fechas,
    'venta': ventas,
    'id_tienda': np.random.randint(1, 50, n_filas),
    'ciudad': ciudades,
    'categoria_producto': categorias
})

# Guardar a CSV
df_gigante.to_csv('ventas_gigantes1.csv', index=False)
print("¡Archivo 'ventas_gigantes1.csv' creado con éxito!")
```
